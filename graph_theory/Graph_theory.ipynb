Dijkstra

```python
G = nx.Graph()
G.add_weighted_edges_from([(1, 2, 6),(1, 3, 3),(1, 4, 1),(2, 5, 1),(3, 2, 2),(3, 4, 2),(4, 6, 10),(5, 4, 6),(5, 6, 4),(5, 7, 3),(5, 8, 6),(6, 5, 10),(6, 7, 2),(7, 8, 4),(9, 5, 2),(9, 8, 3)])
path = nx.dijkstra_path(G, source = 1, target = 8, weight = "weight")
distance = nx.dijkstra_path_length(G, source = 1, target = 8, weight = "weight")

print(path,distance)
```

Floyd

```python
G = nx.Graph()
G.add_weighted_edges_from([(1, 3, 10),
                           (1, 4, 60),
                           (2, 3, 5),
                           (2, 4, 20),
                           (3, 4, 1)
                           ])

dist = dict(nx.floyd_warshall(G, weight = "weight"))

for i in dist:
    print(i, dist[i])

nodes = [1, 2, 3, 4]

D = nx.floyd_warshall_numpy(
    G,
    nodelist = nodes,
    weight = "weight"
)
print(D)
```

$最小生成树，kruskal,prim$

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

G = nx.Graph()
G.add_weighted_edges_from([(0, 1, 2),
                           (0, 2, 1),
                           (0, 3, 3),
                           (0, 4, 4),
                           (0, 5, 4),
                           (0, 6, 2),
                           (0, 7, 5),
                           (0, 8, 4),
                           (1, 2, 4),
                           (1, 8, 1),
                           (2, 3, 1),
                           (3, 4, 1),
                           (4, 5, 5),
                           (5, 6, 2),
                           (6, 7, 3),
                           (7, 8 ,5)])

if not nx.is_connected(G):
    raise ValueError("原图不连通，不存在生成树")

T_kruskal = nx.minimum_spanning_tree(
    G,
    weight = "weight",
    algorithm = "kruskal",
)

T_prim = nx.minimum_spanning_tree(
    G,
    weight = "weight",
    algorithm = "prim",
)

cost_kruskal = T_kruskal.size(weight = "weight")
cost_prim = T_prim.size(weight = "weight")
print("kruskal结果:")
for u, v, weight in sorted(T_kruskal.edges(data = "weight")):
    print(f"{u} -- {v}, 权重 = {weight}")

print("\nprim结果:")
for u, v, weight in sorted(T_prim.edges(data = "weight")):
    print(f"{u} -- {v}, 权重 = {weight}")



labels = {i: f"v{i}" for i in range(9)}

nx.draw(T_kruskal, labels = labels, with_labels=True, node_color = "lightyellow", node_shape="s")

plt.show()

nx.draw(T_prim, labels = labels, with_labels=True, node_color = "lightyellow", node_shape="s")

plt.show()
```

$最大流$

```python
import networkx as nx
from networkx.algorithms.flow import edmonds_karp

G = nx.DiGraph()

edges = [
    ("S","A", 3),
    ("S","B", 2),
    ("A","B", 1),
    ("A","T", 2),
    ("B","T", 3),
]

for u, v, cap in edges:
    G.add_edge(u, v, capacity = cap)

flow_value, flow_dict = nx.maximum_flow(
    G,
    "S",
    "T",
    capacity = "capacity",
    flow_func = edmonds_karp,
)

print("最大流量:",flow_value)

print("\n各边的流量和容量:")

for u, v, data in G.edges(data = True):
    flow = flow_dict[u][v]
    capacity = data["capacity"]
    utilization = flow / capacity

    print(
        f"{u} -> {v}:"
        f"流量 = {flow}"
        f"容量 = {capacity}"
        f"利用率 = {utilization}"
    )

cut_value, partition = nx.minimum_cut(
    G,
    "S",
    "T",
    capacity = "capacity",
    flow_func = edmonds_karp
)

source_side, sink_side = partition

cut_edges = [
    (u, v, G[u][v]["capacity"])
    for u in source_side
    for v in G.successors(u)
    if v in sink_side
]

print("最大流量:",flow_value)
print("各边流量:",flow_dict)
print("最小割容量:",cut_value)
print("源点侧:",source_side)
print("汇点侧:",sink_side)
print("割边:",cut_edges)

assert flow_value == cut_value
```

$最大流练习$

```python
import networkx as nx
from networkx.algorithms.flow import edmonds_karp

G = nx.DiGraph()
edges = [
    ("S","A",4),
    ("S","B",3),
    ("S","C",3),
    ("S","D",2),
    ("S","E",4),
    ("1","T",5),
    ("2","T",4),
    ("3","T",4),
    ("4","T",3),
    ("A","1",1),
    ("B","1",1),
    ("C","1",1),
    ("D","1",1),
    ("E","1",1),
    ("C","2",1),
    ("D","2",1),
    ("A","2*",1),
    ("B","2*",1),
    ("E","2*",1),
    ("2*","2",2),
    ("C","3",1),
    ("B","3",1),
    ("E","3",1),
    ("A","3*",1),
    ("D","3*",1),
    ("3*","3",1),
    ("E","4",1),
    ("A","4*",1),
    ("B","4*",1),
    ("C","4*",1),
    ("D","4*",1),
    ("4*","4",2),
]

for u, v, cap in edges:
    G.add_edge(u, v, capacity = cap)

flow_value, flow_dict = nx.maximum_flow(
    G,
    "S",
    "T",
    capacity = "capacity",
    flow_func = edmonds_karp,
)

print("最大流量:",flow_value)

cut_value, partition = nx.minimum_cut(
    G,
    "S",
    "T",
    capacity = "capacity",
    flow_func = edmonds_karp
)

source_side, sink_side = partition

cut_edges = [
    (u, v, G[u][v]["capacity"])
    for u in source_side
    for v in G.successors(u)
    if v in sink_side
]


for u, v, data in G.edges(data = True):
    flow = flow_dict[u][v]
    capacity = data["capacity"]
    utilization = flow / capacity

    print(
        f"{u} -> {v}:"
        f"容量 = {capacity}"
        f"利用率 = {utilization}"
    )

print("最小割容量:",cut_value)
print("源点侧:",source_side)
print("汇点侧:",sink_side)
print("割边:",cut_edges)

assert flow_value == cut_value
```
