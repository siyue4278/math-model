$最小生成树，kruskal,prim$

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

G = nx.Graph()
G.add_weighted_edges_from([(0, 1, 2),
                           (0, 2, 1),
                           (0, 3, 3),
                           (0, 4, 4),
                           (0, 5, 4),
                           (0, 6, 2),
                           (0, 7, 5),
                           (0, 8, 4),
                           (1, 2, 4),
                           (1, 8, 1),
                           (2, 3, 1),
                           (3, 4, 1),
                           (4, 5, 5),
                           (5, 6, 2),
                           (6, 7, 3),
                           (7, 8 ,5)])

if not nx.is_connected(G):
    raise ValueError("原图不连通，不存在生成树")

T_kruskal = nx.minimum_spanning_tree(
    G,
    weight = "weight",
    algorithm = "kruskal",
)

T_prim = nx.minimum_spanning_tree(
    G,
    weight = "weight",
    algorithm = "prim",
)

cost_kruskal = T_kruskal.size(weight = "weight")
cost_prim = T_prim.size(weight = "weight")
print("kruskal结果:")
for u, v, weight in sorted(T_kruskal.edges(data = "weight")):
    print(f"{u} -- {v}, 权重 = {weight}")

print("\nprim结果:")
for u, v, weight in sorted(T_prim.edges(data = "weight")):
    print(f"{u} -- {v}, 权重 = {weight}")



labels = {i: f"v{i}" for i in range(9)}

nx.draw(T_kruskal, labels = labels, with_labels=True, node_color = "lightyellow", node_shape="s")

plt.show()

nx.draw(T_prim, labels = labels, with_labels=True, node_color = "lightyellow", node_shape="s")

plt.show()
```

$最大流$

```python
import networkx as nx
from networkx.algorithms.flow import edmonds_karp

G = nx.DiGraph()

edges = [
    ("S","A", 3),
    ("S","B", 2),
    ("A","B", 1),
    ("A","T", 2),
    ("B","T", 3),
]

for u, v, cap in edges:
    G.add_edge(u, v, capacity = cap)

flow_value, flow_dict = nx.maximum_flow(
    G,
    "S",
    "T",
    capacity = "capacity",
    flow_func = edmonds_karp,
)

print("最大流量:",flow_value)

print("\n各边的流量和容量:")

for u, v, data in G.edges(data = True):
    flow = flow_dict[u][v]
    capacity = data["capacity"]
    utilization = flow / capacity

    print(
        f"{u} -> {v}:"
        f"流量 = {flow}"
        f"容量 = {capacity}"
        f"利用率 = {utilization}"
    )

cut_value, partition = nx.minimum_cut(
    G,
    "S",
    "T",
    capacity = "capacity",
    flow_func = edmonds_karp
)

source_side, sink_side = partition

cut_edges = [
    (u, v, G[u][v]["capacity"])
    for u in source_side
    for v in G.successors(u)
    if v in sink_side
]

print("最大流量:",flow_value)
print("各边流量:",flow_dict)
print("最小割容量:",cut_value)
print("源点侧:",source_side)
print("汇点侧:",sink_side)
print("割边:",cut_edges)

assert flow_value == cut_value
```

$最大流练习$

```python
import networkx as nx
from networkx.algorithms.flow import edmonds_karp

G = nx.DiGraph()
edges = [
    ("S","A",4),
    ("S","B",3),
    ("S","C",3),
    ("S","D",2),
    ("S","E",4),
    ("1","T",5),
    ("2","T",4),
    ("3","T",4),
    ("4","T",3),
    ("A","1",1),
    ("B","1",1),
    ("C","1",1),
    ("D","1",1),
    ("E","1",1),
    ("C","2",1),
    ("D","2",1),
    ("A","2*",1),
    ("B","2*",1),
    ("E","2*",1),
    ("2*","2",2),
    ("C","3",1),
    ("B","3",1),
    ("E","3",1),
    ("A","3*",1),
    ("D","3*",1),
    ("3*","3",1),
    ("E","4",1),
    ("A","4*",1),
    ("B","4*",1),
    ("C","4*",1),
    ("D","4*",1),
    ("4*","4",2),
]

for u, v, cap in edges:
    G.add_edge(u, v, capacity = cap)

flow_value, flow_dict = nx.maximum_flow(
    G,
    "S",
    "T",
    capacity = "capacity",
    flow_func = edmonds_karp,
)

print("最大流量:",flow_value)

cut_value, partition = nx.minimum_cut(
    G,
    "S",
    "T",
    capacity = "capacity",
    flow_func = edmonds_karp
)

source_side, sink_side = partition

cut_edges = [
    (u, v, G[u][v]["capacity"])
    for u in source_side
    for v in G.successors(u)
    if v in sink_side
]


for u, v, data in G.edges(data = True):
    flow = flow_dict[u][v]
    capacity = data["capacity"]
    utilization = flow / capacity

    print(
        f"{u} -> {v}:"
        f"容量 = {capacity}"
        f"利用率 = {utilization}"
    )

print("最小割容量:",cut_value)
print("源点侧:",source_side)
print("汇点侧:",sink_side)
print("割边:",cut_edges)

assert flow_value == cut_value
```

```python
import networkx as nx
from utils import verify_min_cost_flow
G = nx.DiGraph()

G.add_node("S", demand = -5)
G.add_node("A", demand = 0)
G.add_node("B", demand = 0)
G.add_node("T", demand = 5)

G.add_edge("S", "A", capacity = 4, weight = 2)
G.add_edge("S", "B", capacity = 3, weight = 1)
G.add_edge("A", "B", capacity = 2, weight = 1)
G.add_edge("A", "T", capacity = 3, weight = 3)
G.add_edge("B", "T", capacity = 4, weight = 2)

flow_dict = nx.min_cost_flow(
    G, 
    demand = "demand",
    capacity = "capacity",
    weight = "weight",
)

total_cost = nx.cost_of_flow(
    G,
    flow_dict,
    weight = "weight",
)

print("最小总费用:", total_cost)

print("\n各边流量:")
for u, v, data in G.edges(data = True):
    flow = flow_dict[u][v]

    print(
        f"{u} -> {v}: "
        f"流量={flow}, "
        f"容量={data['capacity']}, "
        f"单位费用={data['weight']}, "
        f"边费用={flow * data['weight']}"
    )

#验证部分

total_cost, flow_dict = nx.network_simplex(G)

verify_min_cost_flow(G, flow_dict)


#灵敏度分析

import networkx as nx


def build_network(sb_capacity: int) -> nx.DiGraph:
    """根据 S->B 的容量建立最小费用流网络。"""

    G = nx.DiGraph()

    # NetworkX：供应为负，需求为正
    G.add_node("S", demand=-5)
    G.add_node("A", demand=0)
    G.add_node("B", demand=0)
    G.add_node("T", demand=5)

    G.add_edge("S", "A", capacity=4, weight=2)
    G.add_edge("S", "B", capacity=sb_capacity, weight=1)
    G.add_edge("A", "B", capacity=2, weight=1)
    G.add_edge("A", "T", capacity=3, weight=3)
    G.add_edge("B", "T", capacity=4, weight=2)

    return G


for sb_capacity in range(1, 7):
    G = build_network(sb_capacity)

    total_cost, flow_dict = nx.network_simplex(
        G,
        demand="demand",
        capacity="capacity",
        weight="weight",
    )

    print(f"\nS->B容量 = {sb_capacity}")
    print(f"最小总费用 = {total_cost}")

    for u, v in G.edges:
        flow = flow_dict[u][v]

        if flow > 0:
            print(
                f"  {u}->{v}: "
                f"流量={flow}, "
                f"单位费用={G[u][v]['weight']}"
            )

total_cost, flow_dict = nx.network_simplex(G)

verify_min_cost_flow(G, flow_dict)
```

```python
import networkx as nx
from utils import verify_min_cost_flow
G = nx.DiGraph()

G.add_node("S")
G.add_node("T")

G.add_edge("S", "2", capacity = 5, weight = 3)
G.add_edge("S", "3", capacity = 3, weight = 6)
G.add_edge("2", "4", capacity = 2, weight = 8)
G.add_edge("3", "2", capacity = 1, weight = 2)
G.add_edge("3", "5", capacity = 4, weight = 2)
G.add_edge("4", "3", capacity = 1, weight = 1)
G.add_edge("4", "5", capacity = 3, weight = 4)
G.add_edge("4", "T", capacity = 2, weight = 10)
G.add_edge("5", "T", capacity = 5, weight = 2)


flow_dict = nx.max_flow_min_cost(
    G, 
    "S",
    "T",
    capacity = "capacity",
    weight = "weight",
)

total_cost = nx.cost_of_flow(
    G,
    flow_dict,
    weight = "weight",
)

max_flow_val = sum(flow_dict["S"].values())
G.nodes["S"]["demand"] = -max_flow_val 
G.nodes["T"]["demand"] = max_flow_val

print("最小总费用:", total_cost)

print("\n各边流量:")
for u, v, data in G.edges(data = True):
    flow = flow_dict[u][v]

    print(
        f"{u} -> {v}: "
        f"流量={flow}, "
        f"容量={data['capacity']}, "
        f"单位费用={data['weight']}, "
        f"边费用={flow * data['weight']}"
    )

#验证部分
verify_min_cost_flow(G, flow_dict)
total_cost, flow_dict = nx.network_simplex(G)
verify_min_cost_flow(G, flow_dict)
```
